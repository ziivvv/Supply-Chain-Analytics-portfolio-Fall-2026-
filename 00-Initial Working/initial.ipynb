```python
import pandas as pd
```

### Corrected Data Loading and Initial Inspection

```python
# Load the Excel file, skipping initial metadata rows and setting the correct header
# Based on inspection, the actual header seems to be in the 4th row (index 3).
df = pd.read_excel('/content/Chenab_Motors_Procurement_Practice_Dataset.xlsx', header=3)

# Display the first 5 rows of the DataFrame to confirm the correct header and data structure
print("First 5 rows of the dataset after correcting header:")
display(df.head())

# Display information about the DataFrame, including data types and non-null values
print("\nDataFrame Info after correcting header:")
df.info()

# Display descriptive statistics for numerical columns
print("\nDescriptive Statistics after correcting header:")
display(df.describe(include='all'))
```

### Data Cleaning and Preparation for Top Suppliers

```python
# Check for missing values in critical columns like 'Vendor Name' and 'Amount'
print("\nMissing values before cleaning (Vendor/Amount):")
display(df[['Vendor', 'Paid Amount (PKR)']].isnull().sum())

# Drop rows where 'Vendor Name' is missing as it's crucial for supplier analysis
df.dropna(subset=['Vendor Name'], inplace=True)

# Fill any remaining missing 'Amount' values with 0 and convert to numeric.
# Coerce errors to NaN and then fill any new NaNs with 0.
df['Amount'] = pd.to_numeric(df['Amount'], errors='coerce')
df['Amount'].fillna(0, inplace=True)

# Verify data types and missing values after cleaning
print("\nMissing values after cleaning (Vendor/Amount):")
display(df[['Vendor Name', 'Amount']].isnull().sum())
print("\nData types after cleaning (Vendor/Amount):")
display(df[['Vendor Name', 'Amount']].dtypes)
```

### Calculations: Identifying Top 3 Suppliers

```python
# Group by 'Vendor Name' and sum the 'Amount' to get total expenditure per supplier
supplier_expenditure = df.groupby('Vendor Name')['Amount'].sum().reset_index()

# Sort the suppliers by total expenditure in descending order
supplier_expenditure_sorted = supplier_expenditure.sort_values(by='Amount', ascending=False)

# Select the top 3 suppliers
top_3_suppliers = supplier_expenditure_sorted.head(3)

print("Top 3 Suppliers by Total Expenditure:")
display(top_3_suppliers)
```

### Visualization: Top 3 Suppliers

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a bar chart for the top 3 suppliers
plt.figure(figsize=(10, 6))
sns.barplot(x='Vendor Name', y='Amount', data=top_3_suppliers, palette='magma')
plt.title('Top 3 Suppliers by Total Procurement Expenditure')
plt.xlabel('Vendor Name')
plt.ylabel('Total Expenditure (Amount)')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Final Summary Table: Top 3 Suppliers

```python
# Display the final summary table of the top 3 suppliers
print("Summary: Top 3 Suppliers and their Total Procurement Expenditure")
display(top_3_suppliers)
```

### Data Cleaning and Preparation

```python
# Check for missing values in critical columns like 'Department Name' and 'Amount'
print("\nMissing values before cleaning:")
display(df[['Department Name', 'Amount']].isnull().sum())

# Drop rows where 'Department Name' is missing as it's crucial for our analysis
df.dropna(subset=['Department Name'], inplace=True)

# Fill any remaining missing 'Amount' values with 0 or a suitable median/mean if appropriate.
# For procurement data, 0 is often a reasonable default for missing amounts if they imply no cost.
# However, if 'Amount' is critical and missing often, further investigation might be needed.
# For this analysis, we'll fill with 0 to ensure numerical operations don't fail.
df['Amount'].fillna(0, inplace=True)

# Convert 'Amount' column to numeric, coercing errors to NaN and then filling with 0 if any new NaNs appear
df['Amount'] = pd.to_numeric(df['Amount'], errors='coerce')
df['Amount'].fillna(0, inplace=True)

# Verify data types and missing values after cleaning
print("\nMissing values after cleaning:")
display(df[['Department Name', 'Amount']].isnull().sum())
print("\nData types after cleaning:")
display(df[['Department Name', 'Amount']].dtypes)
```

### Calculations: Identifying Top Departments

```python
# Group by 'Department Name' and sum the 'Amount' to get total expenditure per department
department_expenditure = df.groupby('Department Name')['Amount'].sum().reset_index()

# Sort the departments by total expenditure in descending order
department_expenditure_sorted = department_expenditure.sort_values(by='Amount', ascending=False)

# Select the top 3 departments
top_3_departments = department_expenditure_sorted.head(3)

print("Top 3 Departments by Total Expenditure:")
display(top_3_departments)
```

### Visualization: Top 3 Departments

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a bar chart for the top 3 departments
plt.figure(figsize=(10, 6))
sns.barplot(x='Department Name', y='Amount', data=top_3_departments, palette='viridis')
plt.title('Top 3 Departments by Total Procurement Expenditure')
plt.xlabel('Department Name')
plt.ylabel('Total Expenditure (Amount)')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Final Summary Table

```python
# Display the final summary table of the top 3 departments
print("Summary: Top 3 Departments and their Total Procurement Expenditure")
display(top_3_departments)
```
